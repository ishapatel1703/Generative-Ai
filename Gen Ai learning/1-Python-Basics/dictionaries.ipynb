Dictionaries are unordered collection of items, They store data in key value pairs, Key must be unique and immutable(eg, string, numbers or tuples),while values can be of any  type

```python
#creating dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"Isha", "age":32, "grade":24}
student
```

```python
#Error
#single key is always used
student={"name":"Isha", "age":32, "name":"Ishaa"}#wrong
student
```

```python
#Accessing dictionaries elements
print(student['age'])
print(student['name'])
#Accessing using get() method
print(student.get('age'))
print(student.get('name'))
print(student.get('lastname', "Not found"))
```

```python
#Modifying Dictionary elements
##Dictionaries are mutable so you can add, uodate or delete elements
print(student)
```

```python
student["age"]=22
print(student)
student["address"]="Jabalpur"
student["grade"]="A"
print(student)
```

```python
# delete key and value pair
del student['grade']
print(student)
```

```python
#Dictionary method

keys=student.keys()
print(keys)
values=student.values()
print(values)
items=student.items()
print(items)
```

```python
#shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="IshaP"
print(student)
print(student_copy)#both will be change
```

```python
student_copy1=student.copy() #shalow copy right way
print(student)
print(student_copy1)
```

```python
student["name"]="IshaaP"
print(student)
print(student_copy1) #both are not same
```

```python
##Iterating over Dictionaries
##you can use loops to iterate over dictionaries, key value or items

#iterating over key
for keys in student.keys():
    print(keys)
```

```python
#iterating over key
for values in student.values():
    print(values)
```

```python
#iterating over key value pairs
for keys,value in student.items():
    print(f"{keys}:{values}")
```

```python
#Nested Dictionaries
students={
    "student1":{"Name":"Ishaaa", "Age":22},
    "student2":{"Name":"Patelll", "Age":22}
}
print(students)
```

```python
#Access Nested Dictionaries Elements
print(students["student1"]["Name"])
print(students["student2"]["Name"])

```

```python
students.items()
```

```python
#Iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}, {student_info}")
    for keys,values in student_info.items():
        print(f"{keys},{values}")
              
            
```

```python
#Dictionaries Comprehension
squares={x:x**2 for x in range(5)}
print(squares)

```

```python
#Condition Dictionaries Comprehension
even_squares={x:x**2 for x in range(10) if x%2==0}
print(even_squares)

```

```python
numbers=[1,2,2,2,2,3,2,3,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1

    else:
        frequency[number]=1

print(frequency)
```

```python
#merge 2 dictionaries into one
dict1={"a":1, "b":2}
dict2={"b":3, "c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
