```python
##Data Serialization
#convert json into a str or dict(parced data) form

import json
data={'name':'Isha', 'age':25}

json_str= json.dumps(data)
print(json_str)
print(type(json_str))


parsed_data= json.loads(json_str)
print(parsed_data)
print(type(parsed_data))


#whenever we get json in the form of an api whcih is in the str format so we just load that perticular str json to convert it into a dict form 
```

```python
import csv

with open("example.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age"])
    writer.writerow(["Isha", 32])

print("File written!")

with open("example.csv", "r") as file:
    print("Reading...")
    reader = csv.reader(file)

    for row in reader:
        print(row)

print("Done")
```
