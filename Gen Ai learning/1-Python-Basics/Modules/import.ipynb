In python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(24))
print(sqrt(34))
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
#for import all the fun
from  math import *
print(sqrt(16))
print(pi)
```

```python
from package.maths import addition
addition(8,3)
```

```python
#Also you can write
from package import maths
addition(8,3)
```

```python
from package.maths import subtraction
subtraction(8,3)
```

```python
from package import maths
maths.subtraction(8,3)
```
