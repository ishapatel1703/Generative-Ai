```python
to_do_list=["buy Groceries", "Clean the house", "Pay bills"]

#Adding to task
to_do_list.append("Schedule meeting")
to_do_list.append("Go for a run")

#removing a complete task
to_do_list.remove("Clean the house")

#Checking if task is in the list
if "Pay bills" in to_do_list:
    print("Don't forget to pay the bill")

print("Remaining to do list ")
for task in to_do_list:
    print(f"-{task}")
```

2.Organizing a Student grades

```python
#Organizing a Student grades
grades=[88,85,95,99,96]

#Adding a new grade
grades.append(95)

#Calculate the avg marks 
avg_grade= sum(grades) / len(grades)
print(f"Average Grade: {avg_grade:.2f}")

#Finding the highest and lowest grades
highest_grade=max(grades)
lowest_grade=min(grades)
print(f"highest_grade: {highest_grade}")
print(f"lowest_grade: {lowest_grade}")
```

3.Manage Inventory
