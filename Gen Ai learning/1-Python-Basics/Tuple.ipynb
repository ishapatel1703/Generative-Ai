```python
#packing and unpacking tuples
#packing
packed_tuple=1,"Hello",3.14
print(packed_tuple)
```

```python
#uppacked tuple
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
#Unpacking with *
numbers=(1,2,3,4,5)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
#Nested tuple
##Nested list
lst=[[1,2,3,4,5],[6,7,8,9],[1,"hi",3.14,True]]
lst[0]  #[1,2,3,4,5]
```

```python
##Nested list
lst=[[1,2,3,4,5],[6,7,8,9],[1,"hi",3.14,True]]
lst[0][2]#3
```

```python
##Nested list
lst=[[1,2,3,4,5],[6,7,8,9],[1,"hi",3.14,True]]
lst[0][0:2]#[1,2,3]
```

```python
lst=[[1,2,3,4,5],[6,7,8,9],(1,"hi",3.14,True)]
lst[2][0:2]#(1,"hi",3.14)
```
