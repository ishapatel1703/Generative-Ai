The map ()  function applies a given function to all items in an input list(or any other iterable) and returns a map object (an iterator). This is a perticularly useful for transforming data in a list comprehensively

```python
def square(x):
    return x*x
square(4)
```

```python
numbers=[1,2,3,4,5,6]

list(map(square, numbers))
```

```python
#with lambda function
numbers=[1,2,3,4,5,6]
list(map(lambda x:x*x, numbers))
```

```python
#map multiple iterable
number1=[1,2,3,4]
number2=[5,6,7,8]

added_numbers=list(map( lambda x,y:x+y,number1,number2))
print(added_numbers)
```

```python
#map() to convert a list of string to integers
#Use map to convert strings to integers

str_numbers=['1','2','3','4']
int_numbers=list(map(int,str_numbers))
print(int_numbers)
```

```python
words=['apple','banana','cherry']
upper_words=list(map(str.upper,words))
print(upper_words)
```

```python
def get_name(person):
    return person['name']

people =[
    {'name':'Isha','age':'22'},
    {'name':'Ashi','age':'23'}
]

list(map(get_name,people))
```

The map() function is a powerful tool for applying transformations to iterable data structures.
