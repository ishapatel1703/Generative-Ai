A function is a block of code that performs a specific task, functions help in organizing code,reusing and readability.


```python
#Syntax
def function_name(parameters:):
    """Docstring"""
    #function body
    return expression
```

```python
def  even_or_odd(num):
    """this funtion finds even or odd"""
    if num%2==0:
        print("Num is even")
    else:
        print("Num is odd")
```

```python
#call a function
even_or_odd(28)
```

```python
#multiple parameters
def add(a,b):
    return a+b
result=add(2,4)
print(result)
```

```python
#Default parameters
def greet(name="Guest"):
    print(f"Hello {name} welcome to the paradise")

greet()
```

```python
#Default parameters with parameter
def greet(name="Guest"):
    print(f"Hello {name} welcome to the paradise")

greet("Isha")
```

```python
#Variable Length argument
##Positional argument

def  print_numbers(*args): #it could be anything like *Isha, *udemy etc
     for number in args:
          print(number)
```

```python
print_numbers(1,2,3,4,5,6,"Hi")
```

```python
# keyword argument
def print_details(**kwargs):#it could be anything like *Isha, *udemy etc
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Isha",age=22,grade="A+")
```

```python
def print_details(*args,**kwargs):

    for val in args:
              print(f"{val}")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"hi",name="Isha",age=22,grade="A+")
```
