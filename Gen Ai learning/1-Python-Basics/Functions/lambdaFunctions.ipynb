Lambda Functions are small anonymous functions defined using the lambda keywords. They can  have any number of arguments  but only one expression. They are  commonly used for short operations or as arguement to higher-order functions

```python
#syntax
lambda arguments : expression
```

```python
addition= lambda a,b : a+b
type(addition)
```

```python
addition(2,3)
```

```python
even1 =lambda num :num%2==0
even1(13)
```

```python
addition1=lambda x,y,z : x+y+z
addition1(2,3,4)
```

```python
## map() applies a function to all items in a list
numbers=[1,2,3,4,5,6]

##normal way
def square(numbers):
    return numbers**2
square(2)
```

```python
#using map
list(map(lambda x:x**2, numbers))
```
